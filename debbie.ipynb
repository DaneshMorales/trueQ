```python
import trueq as tq
import trueq.math as tqm
import trueq.simulation as tqs
import numpy as np
```

```python
tqm.Superop.from_unitary(tq.Gate.h.mat).plot_ptm()
```

```python
noisy_sim = tq.Simulator().add_overrotation(single_sys=0.1, match= tqs.GateMatch(tq.Gate.h))
```

```python
cb_circs = tq.make_cb(cycles= tq.Cycle({0:tq.Gate.h}), n_random_cycles= [2,4,8,16], n_circuits = 1000, n_decays=4)
```

```python
noisy_sim.run(cb_circs, n_shots= np.inf)
```
